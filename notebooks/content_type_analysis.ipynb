# **Content Type Analysis Dashboard**

**Loading the Cleaned Dataset**

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("Cleaned_Dataset.csv")
data.head(10)
```

```python
data.info()
```

####**Movies vs TV Shows**

```python
content_type_counts = data['Type'].value_counts()
print(content_type_counts)
```

**Visualizations for content distribution**

```python
##Bar Charts for Comparing Categories(Best for comparison)
content_type_counts.plot(kind = 'bar', figsize=(8,5))

plt.title('Netflix Content Distribution')
plt.xlabel('Content Type')
plt.ylabel('Value Counts')
plt.show()
```

```python
movies = (data['Type'] == 'Movie').sum()
tv_shows = (data['Type'] == 'TV Show').sum()

total_content  = movies + tv_shows
print('Total Movies: ' , movies)
print('Total TV Shows: ' , tv_shows)
print('Total Content: ',total_content)
movies_ratio = (movies/total_content)*100.00
tv_shows_ratio = (tv_shows/total_content)*100.00
print('\n')
print(f'Movies covered {movies_ratio:.2f}% of Total Content')
print(f'TV Shows covered {tv_shows_ratio:.2f}% of Total Content')
```

```python
##Pie Charts (best for Parts of a Numerical Whole)
content_type_counts.plot(kind = 'pie', autopct = '%.2f%%' ,figsize=(8,8))

plt.title('Netflix Movies Vs TV Shows')
plt.show()
```

**Comparison of Content Proportions**

```python
print("Content Proportions")
print('Total Movies: ' , movies)
print('Total TV Shows: ' , tv_shows)

print('\n')
if movies > tv_shows:
    print("Movies are more common than TV Shows.")
elif tv_shows > movies:
    print("TV Shows are more common than Movies.")
else:
    print("Movies and TV Shows have the equal proportion.")
```

**Key findings**

```python
print('\n*************Key Findings**************\n')
print(f'1. Netflix contains {movies} Movies.')
print(f'2. Netflix contains {tv_shows} TV Shows.')
print(f'3. Movies covered {movies_ratio:.2f}% of Total Content')
print(f'4. TV Shows covered {tv_shows_ratio:.2f}% of Total Content')
```
