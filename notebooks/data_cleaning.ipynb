# **Netflix Data Cleaning & Preparation**

**Data Loading**

```python
import pandas as pd

data = pd.read_csv("Dataset.csv")
data.head()
```

**Data Inspection**

```python
data.info()
print('\n',data.shape)
print('\n',data.columns)
```

**Missing Values Check**

```python
data.isnull().sum()
```

The dataset does not contain missing values .

**Duplicates Check**

```python
print(data.duplicated().sum())
```

The dataset does not contain duplicated records

**Columns Standardization**

```python
data.columns
```

```python
##Remove unnessary spaces
data.columns = data.columns.str.strip()
data['date_added'] = data['date_added'].str.strip()
print('\n',data.columns)
print(data.head(10))
##Convert to standard title format
data.columns = data.columns.str.title()
print('\n',data.columns)
```

**Value Standardization**

```python
##Check for unique values in each column
print(data['Country'].unique())
print(data['Type'].unique())
print(data['Rating'].unique())
```

All columns' values are already clean and standardized

**Exporting the cleaned Dataset**

```python
data.to_csv('Cleaned_Dataset.csv' , index=False)
print("Cleaned Data Exported")
```

**Conclusion:**

The Netflix dataset was inspected for data quality issues. No missing values or duplicate records were found, and the checked categorical fields such as Type, Country, and Rating were already standardized. Therefore, no corrective data cleaning operations were required. The validated dataset was exported as Cleaned_Dataset.csv for analysis.
