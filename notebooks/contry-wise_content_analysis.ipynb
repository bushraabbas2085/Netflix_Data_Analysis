# **Country-Wise Netflix Content Analysis**

**Dataset Loading**

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("Cleaned_Dataset.csv")
data.head(10)
```

```python
data.info()
```

**Extracting and Cleaning Country Information**

```python
country_info = data[['Title','Country']]
print(country_info)
```

Country Information is already cleaned

**Countries' Content Count**

```python
country_content_count = data.groupby('Country')['Title'].count()
print(country_content_count)
```

**Identifying top Content Producing Countries**

```python
top_countries = country_content_count.sort_values(ascending=False).head(10)
top_countries_data = top_countries.reset_index()
top_countries_data.columns = ['Country' , 'Content_Count']
print("Top content producing countries\n")
top_countries_data.head(10)
```

**Charts and rankings Creation**

```python
top_countries_data['Rank'] = range(1, len(top_countries)+1)
print('Top 10 Countries by Content\n')
top_countries_data[['Country' , 'Content_Count' , 'Rank']]
```

Visualizations of Distribution of Content across top Countries

```python
##Horizontal Bar Chart
##Comparison of Distribution of Content across top Countries
plt.figure(figsize=(10, 6))

plt.bar(
    top_countries_data["Country"],
    top_countries_data["Content_Count"]
)
plt.title("Top 10 Countries by Content")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
##Pie Chart

top_countries_data['Content_Count'].plot(kind = 'pie',
                                         autopct = '%.2f%%' ,
                                         figsize=(8,8),
                                         labels=top_countries_data['Country'])

plt.title('Top Countries Content Ratio')
plt.show()
```
